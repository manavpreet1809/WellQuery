```python
from pyspark.sql import SparkSession
from helpers.cleaning import clean_and_sentence_split
```

```python
spark = SparkSession.builder.getOrCreate()
spark.sql("USE CATALOG workspace")
spark.sql("USE SCHEMA med")
```

```python
raw_df = spark.table("workspace.med.raw_data")
```

```python
df_docs_clean = clean_and_sentence_split(raw_df, min_len=50)
display(df_docs_clean.limit(5))
```

```python
(
    df_docs_clean
      .write
      .format("delta")
      .mode("overwrite")
      .option("overwriteSchema", "true")
      .saveAsTable("workspace.med.docs_clean")
)
```
