```python
import sys
import os

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
if not os.path.isdir(os.path.join(PROJECT_ROOT, "helpers")):
    raise RuntimeError("Run from the imported Databricks/notebooks directory")

# remove it if it exists so we can reinsert at position 0
if PROJECT_ROOT in sys.path:
    sys.path.remove(PROJECT_ROOT)

# insert project root at the front so "helpers" resolves correctly
sys.path.insert(0, PROJECT_ROOT)

print("Bootstrap loaded. Project root first in sys.path:", sys.path[0])
```
