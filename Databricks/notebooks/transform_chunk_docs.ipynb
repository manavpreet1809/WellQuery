```python
from pyspark.sql import SparkSession
from helpers.chunking import build_doc_chunks
```

```python
spark = SparkSession.builder.getOrCreate()
spark.sql("USE CATALOG workspace")
spark.sql("USE SCHEMA med")
```

```python
docs_clean_df = spark.table("workspace.med.docs_clean")
```

```python
doc_chunks_df = build_doc_chunks(docs_clean_df, chunk_size=800)
display(doc_chunks_df.limit(10))
```

```python
(
    doc_chunks_df
      .write
      .format("delta")
      .mode("overwrite")
      .option("overwriteSchema", "true")
      .saveAsTable("workspace.med.doc_chunks")
)
```
